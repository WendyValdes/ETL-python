# 1. EXTRACT (KAGGLEHUB)

```python
import kagglehub
import pandas as pd
import random
import os
```

# Download dataset

```python
path = kagglehub.dataset_download("nilaychauhan/world-bank-datasets")

print("📁 Caminho do dataset:", path)
```

# Lista arquivos disponíveis

```python
files = os.listdir(path)
print("\n📂 Arquivos disponíveis:", files)
```

# Vamos pegar um arquivo CSV (ajuste se necessário)

```python
file_path = [f for f in files if f.endswith(".csv")][0]
full_path = os.path.join(path, file_path)
```

# Carrega dados

```python
df = pd.read_csv(full_path, header=2)

print("\n📊 Preview do dataset:")
print(df.head())
```

# Preparação dos "usuários"

# Seleciona algumas colunas disponíveis (adaptável conforme dataset)

```python
df = df.head(5)
```

# Criando estrutura padrão

```python
users = []

for i, row in df.iterrows():
    users.append({
        "id": i,
        "name": str(row.iloc[0])[:10],  # usa primeira coluna como "nome"
        "info": str(row.iloc[1])[:20],
        "news": []
    })

print("\n📥 Dados extraídos:")
print(users)
```

# 2. TRANSFORM (IA SIMULADA)


```python
def generate_ai_news(user):
    mensagens = [
        f"{user['name']}, investir é essencial para crescimento econômico.",
        f"{user['name']}, dados mostram: investir gera mais estabilidade.",
        f"{user['name']}, comece a investir e fortaleça seu futuro.",
        f"{user['name']}, economia forte começa com bons investimentos.",
        f"{user['name']}, decisões inteligentes geram riqueza."
    ]
    return random.choice(mensagens)

for user in users:
    news = generate_ai_news(user)
    user['news'].append({
        "icon": "📊",
        "description": news
    })

print("\n🤖 Dados transformados:")
print(users)
```

# 3. LOAD (SALVAR RESULTADO)

# Normaliza JSON → DataFrame

```python
final_df = pd.json_normalize(users, record_path='news', meta=['id', 'name', 'info'])
```

# Salva CSV

```python
output_file = "resultado_worldbank_etl.csv"
final_df.to_csv(output_file, index=False)

print(f"\n💾 Arquivo salvo: {output_file}")
```
